```python
import requests
import pprint
from getpass import getpass
```

```python
# APIキーを指定
API_KEY = getpass('APIキーを入力してください:')
```

```python
# 駅名を指定
station_name = input('駅名を入力してください:')
```

```python
# 駅の緯度・経度を取得
geocode_url = 'https://maps.googleapis.com/maps/api/geocode/json'

geocode_params = {
    "address": station_name,
    "key": API_KEY,
    "language": "ja"
}

response = requests.get(geocode_url, params=geocode_params)
geocode_data = response.json()

# Geocoding APIの結果を確認
if geocode_data["status"] != "OK":
    print("駅の位置情報を取得できませんでした。")
    print("ステータス:", geocode_data["status"])
    exit()

# 緯度・経度を取得
location = geocode_data["results"][0]["geometry"]["location"]

latitude = location["lat"]
longitude = location["lng"]

print()
print("【駅の位置情報】")
print("駅名:", station_name)
print("緯度:", latitude)
print("経度:", longitude)
```

```python
# 半径500m以内のレストランを取得

nearby_url = "https://maps.googleapis.com/maps/api/place/nearbysearch/json"

nearby_params = {
    "location": f"{latitude},{longitude}",
    "radius": 500,
    "type": "restaurant",
    "key": API_KEY,
    "language": "ja"
}

response = requests.get(nearby_url, params=nearby_params)
nearby_data = response.json()

# Places APIの結果を確認
if nearby_data["status"] not in ["OK", "ZERO_RESULTS"]:
    print("レストラン情報を取得できませんでした。")
    print("ステータス:", nearby_data["status"])
    exit()
```

```python
# レストラン情報を出力

print()
print("========================================")
print(f"「{station_name}」から半径500m以内のレストラン")
print("========================================")

restaurants = nearby_data.get("results", [])

if len(restaurants) == 0:
    print("レストランが見つかりませんでした。")
else:
    for i, restaurant in enumerate(restaurants, start=1):

        # レストラン名
        name = restaurant.get("name", "情報なし")

        # 評価
        rating = restaurant.get("rating", "評価なし")

        # 住所
        address = restaurant.get("vicinity", "住所なし")

        print()
        print(f"{i}. {name}")
        print(f"   評価：{rating}")
        print(f"   住所：{address}")
```
